```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
# Show Items With dir 
dir(iris)
```

```python
iris.data[:,0]
```

```python
scatter =plt.scatter(iris.data[:,0],iris.data[:,1],c=iris.target)
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.legend(scatter.legend_elements()[0],iris.target_names,loc='lower right',title="Classes")
plt.show()
```

```python
scatter =plt.scatter(iris.data[:,2],iris.data[:,3],c=iris.target)
plt.xlabel(iris.feature_names[2])
plt.ylabel(iris.feature_names[3])
plt.legend(scatter.legend_elements()[0],iris.target_names,loc='lower right',title="Classes")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test, y_train,y_test = train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
```

```python
len(x_train),len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict([iris.data[50]])
```

```python
iris.data[50],iris.target[50]
```

```python
from sklearn.metrics import confusion_matrix
y_preq = model.predict(x_test)
cm = confusion_matrix(y_test,y_preq)
cm
```

```python
import seaborn as sn
sn.heatmap(cm,annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
plt.show()
```
