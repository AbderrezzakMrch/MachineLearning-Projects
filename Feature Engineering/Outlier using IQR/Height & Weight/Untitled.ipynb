```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('height_weight.csv')
df.head()
```

```python
df.shape
```

```python
plt.hist((df.gender,df.height),bins=2)
plt.ylabel('Count height')
plt.xlabel('Gender')
plt.show()
```

```python
plt.hist((df.gender,df.weight),bins=2)
plt.ylabel('Count weight')
plt.xlabel('Gender')
plt.show()
```

```python
Q1 = df.weight.quantile(0.25)
Q3 = df.weight.quantile(0.75)
Q1,Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_lim = Q1 - 1.5 * IQR
lower_lim
```

```python
upper_lim = Q3 + 1.5 * IQR
upper_lim
```

```python
df[(df.weight<lower_lim)|(df.weight>upper_lim)]
```

```python
Q1 = df.height.quantile(0.25)
Q3 = df.height.quantile(0.75)
Q1,Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_lim = Q1 - 1.5 * IQR
lower_lim
```

```python
upper_lim = Q3 + 1.5 * IQR
upper_lim
```

```python
df[(df.height<lower_lim)|(df.height>upper_lim)]
```
