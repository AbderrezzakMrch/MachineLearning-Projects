```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import math
```

```python
def gradient_descent(x,y):
    m_curr = b_curr = 0
    iteration = 100000
    n = len(x)
    learning_rate = 0.0002
    cost_previous = 0

    for i in range(iteration):
        y_predicted = m_curr * x + b_curr
        mse = (1/n) * sum([(val)**2 for val in (y-y_predicted)])
        mb = -(2/n) * sum(x*(y-y_predicted))
        bd = -(2/n) * sum(y-y_predicted)

        m_curr = m_curr - learning_rate * mb
        b_curr = b_curr - learning_rate * bd

        if math.isclose(mse,cost_previous,rel_tol=1e-20):
            break
        cost_previous = mse

        print(f'm{m_curr} b {b_curr} mse {mse} iteration {iteration}')

        return m_curr,b_curr
```

```python
def sklearn_model():
    df = pd.read_csv('test_scores.csv')
    reg = LinearRegression()
    reg.fit(df[['math']],df.cs)
    return reg.coef_,reg.intercept_
```

```python
df = pd.read_csv('test_scores.csv')
x = np.array(df.math)
y = np.array(df.cs)
```

```python
m,b = gradient_descent(x,y)
print(f'using gradient descent coef is {m} and intersepts {b}')
```

```python
m_sk,b_sk = sklearn_model()
print(f'using sklearn model coef is {m_sk} and intersepts {b_sk}')
```
